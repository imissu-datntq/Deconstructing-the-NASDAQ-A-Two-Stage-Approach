# 05 Forecast Data and Splits

Goal: prepare monthly NASDAQ target series and define train/validation/test splits for ARIMA/SARIMA.

```python
from pathlib import Path

import pandas as pd

from nasdaq_svar.config import load_config
from nasdaq_svar.forecast import _read_target_series, _split_series

cwd = Path.cwd().resolve()
PROJECT_ROOT = cwd if (cwd / "configs").exists() else cwd.parent
cfg = load_config(PROJECT_ROOT / "configs/forecast.yaml")
forecast_cfg = cfg["forecast"]
input_csv = PROJECT_ROOT / forecast_cfg["input_csv"]
series = _read_target_series(input_csv, forecast_cfg.get("target_column"))
series.tail()
```

```python
split = _split_series(
    series,
    train_ratio=forecast_cfg["evaluation"]["train_ratio"],
    valid_ratio=forecast_cfg["evaluation"]["valid_ratio"],
    min_train_size=forecast_cfg["evaluation"]["min_train_size"],
)

print({
    "train": len(split.train),
    "valid": len(split.valid),
    "test": len(split.test),
})
```
